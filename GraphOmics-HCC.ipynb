```python
import pandas as pd

platform = pd.read_csv(
    "data/raw/geo/GSE64041_GPL6244_platform.csv"
)

def clean_symbol(value):
    if pd.isna(value):
        return None

    s = str(value).strip()

    if not s or s.lower() in {
        "---",
        "na",
        "nan",
        "null",
        "none"
    }:
        return None

    return s


def symbol_from_gene_assignment(value):

    if pd.isna(value):
        return None

    symbols = []

    for assignment in str(value).split("///"):

        parts = [
            x.strip()
            for x in assignment.split("//")
        ]

        if len(parts) < 2:
            continue

        sym = clean_symbol(parts[1])

        if sym is not None:
            symbols.append(sym)

    unique = list(
        dict.fromkeys(symbols)
    )

    return (
        unique[0]
        if len(unique) == 1
        else None
    )


platform["parsed_symbol"] = (
    platform["gene_assignment"]
    .apply(symbol_from_gene_assignment)
)

print("Total platform rows:",
      len(platform))

print(
    "Rows with usable gene symbol:",
    platform["parsed_symbol"].notna().sum()
)

print(
    "Mapping percentage:",
    round(
        platform["parsed_symbol"]
        .notna()
        .mean() * 100,
        2
    ),
    "%"
)

print("\nExample mappings:")

print(
    platform[
        ["ID",
         "gene_assignment",
         "parsed_symbol"]
    ]
    .dropna(subset=["parsed_symbol"])
    .head(20)
    .to_string(index=False)
)
```

```python
from pathlib import Path

p = Path("src/model.py")
text = p.read_text()

old_l1 = '''    def l1(self) -> torch.Tensor:
        return (self.weight * self.mask).abs().sum()
'''

new_l1 = '''    def l1(self) -> torch.Tensor:
        # Mean absolute weight over biologically active memberships only.
        # This makes the regularisation strength independent of pathway-mask size.
        active = (self.weight * self.mask).abs()
        return active.sum() / self.mask.sum().clamp(min=1.0)
'''

old_smooth = '''    B, G = z.shape
    flat = z.t()                                   # (G, B)
    az = torch.sparse.mm(A, flat).t()              # (B, G)
    return ((z * (z - az)).sum(dim=1)).mean()
'''

new_smooth = '''    B, G = z.shape
    flat = z.t()                                   # (G, B)
    az = torch.sparse.mm(A, flat).t()              # (B, G)

    # Mean Dirichlet energy per gene rather than a raw sum over all genes.
    # This prevents regularisation magnitude from growing with graph size.
    return ((z * (z - az)).sum(dim=1) / G).mean()
'''

changed = False

if old_l1 in text:
    text = text.replace(old_l1, new_l1)
    changed = True
    print("✅ L1 regulariser normalized")
elif "regularisation strength independent of pathway-mask size" in text:
    print("✅ L1 already normalized")
else:
    print("⚠️ L1 block not found")

if old_smooth in text:
    text = text.replace(old_smooth, new_smooth)
    changed = True
    print("✅ Graph smoothness normalized")
elif "Mean Dirichlet energy per gene" in text:
    print("✅ Graph smoothness already normalized")
else:
    print("⚠️ Smoothness block not found")

if changed:
    p.write_text(text)
```

```python
from pathlib import Path
import re

p = Path("src/04_train_eval.py")

# Backup current working version
backup = Path("src/04_train_eval_before_nested_fix.py")
if not backup.exists():
    backup.write_text(p.read_text())

text = p.read_text()

if "untouched outer-fold OOF CV" in text:
    print("✅ Nested evaluation patch already applied.")

else:
    # ------------------------------------------------------------
    # Imports
    # ------------------------------------------------------------
    text = text.replace(
        "from sklearn.model_selection import StratifiedKFold",
        "from sklearn.model_selection import StratifiedKFold, train_test_split"
    )

    if "from sklearn.linear_model import LogisticRegression" not in text:
        text = text.replace(
            "from sklearn.model_selection import StratifiedKFold, train_test_split",
            "from sklearn.model_selection import StratifiedKFold, train_test_split\n"
            "from sklearn.linear_model import LogisticRegression\n"
            "from sklearn.metrics import roc_auc_score"
        )

    # ------------------------------------------------------------
    # Calibration helpers
    # ------------------------------------------------------------
    old = '''def ensemble_predict(models, X, A, edge_index, device):
    return np.mean([predict(m, X, A, edge_index, device) for m in models], axis=0)
'''

    new = '''def ensemble_predict(models, X, A, edge_index, device):
    return np.mean(
        [predict(m, X, A, edge_index, device) for m in models],
        axis=0
    )


def fit_platt_scaler(y, p):
    """Development-only Platt scaling."""
    p = np.clip(
        np.asarray(p, dtype=float),
        1e-6,
        1 - 1e-6
    )

    z = np.log(
        p / (1 - p)
    ).reshape(-1, 1)

    lr = LogisticRegression(
        C=1.0,
        solver="lbfgs",
        max_iter=2000
    )

    lr.fit(
        z,
        np.asarray(y, dtype=int)
    )

    return {
        "intercept": float(lr.intercept_[0]),
        "slope": float(lr.coef_[0][0]),
    }


def apply_platt_scaler(p, cal):
    p = np.clip(
        np.asarray(p, dtype=float),
        1e-6,
        1 - 1e-6
    )

    z = np.log(p / (1 - p))

    eta = (
        cal["intercept"]
        + cal["slope"] * z
    )

    eta = np.clip(
        eta,
        -40,
        40
    )

    return 1.0 / (
        1.0 + np.exp(-eta)
    )


def calibrated_ensemble_predict(
    models,
    calibrators,
    X,
    A,
    edge_index,
    device
):
    probs = []

    for model, cal in zip(
        models,
        calibrators
    ):
        raw = predict(
            model,
            X,
            A,
            edge_index,
            device
        )

        probs.append(
            apply_platt_scaler(
                raw,
                cal
            )
        )

    return np.mean(
        probs,
        axis=0
    )
'''

    if old not in text:
        raise RuntimeError(
            "Could not locate ensemble_predict block."
        )

    text = text.replace(old, new)

    # ------------------------------------------------------------
    # Replace run_main
    # ------------------------------------------------------------
    start = text.index(
        "def run_main(cfg, device, save_artifacts=True):"
    )

    end = text.index(
        "\n\n# --------------------------------------------------------------------------\n\n\ndef run_baselines",
        start
    )

    new_run_main = r'''def run_main(cfg, device, save_artifacts=True):
    X, samples, coverage, edge_index, edge_weight, M, H = load_inputs(cfg)

    Xt = to_tensor(
        X,
        samples,
        coverage,
        cfg.get(
            "features",
            "expression_plus_coverage"
        )
    )

    y = samples["tumour"].values.astype(int)
    cohort = samples["cohort"].values

    dev = cfg["cohorts"]["development"]

    A = normalise_adjacency(
        edge_index,
        Xt.shape[1],
        edge_weight
    )

    dev_idx = np.flatnonzero(
        cohort == dev
    )

    print(
        f"development cohort {dev}: "
        f"{len(dev_idx)} samples "
        f"({y[dev_idx].sum()} tumour)"
    )

    # ---------------------------------------------------------
    # STRICT OUTER CV
    # Outer validation fold is never used for early stopping.
    # ---------------------------------------------------------
    outer = StratifiedKFold(
        cfg["train"]["outer_folds"],
        shuffle=True,
        random_state=cfg["seed"]
    )

    inner_frac = float(
        cfg["train"].get(
            "inner_val_fraction",
            0.20
        )
    )

    fold_models = []
    fold_calibrators = []

    oof_pred = []
    oof_true = []
    oof_sample_idx = []
    histories = []

    t0 = time.time()

    for k, (tr, va) in enumerate(
        outer.split(
            dev_idx,
            y[dev_idx]
        )
    ):

        print(
            f"  outer fold "
            f"{k + 1}/"
            f"{cfg['train']['outer_folds']}"
        )

        outer_tr_i = dev_idx[tr]
        va_i = dev_idx[va]

        # Inner split only from outer-training samples
        fit_i, inner_i = train_test_split(
            outer_tr_i,
            test_size=inner_frac,
            stratify=y[outer_tr_i],
            random_state=(
                cfg["seed"]
                + 1000
                + k
            )
        )

        model, inner_auc, hist = train_one(
            Xt[fit_i],
            y[fit_i],

            Xt[inner_i],
            y[inner_i],

            A,
            edge_index,
            M,
            H,
            cfg,

            cfg["seed"] + k,
            device
        )

        # ---------------------------------------------
        # Development-only probability calibration
        # ---------------------------------------------
        p_inner_raw = predict(
            model,
            Xt[inner_i],
            A,
            edge_index,
            device
        )

        cal = fit_platt_scaler(
            y[inner_i],
            p_inner_raw
        )

        # ---------------------------------------------
        # Untouched outer hold-out
        # ---------------------------------------------
        p_va_raw = predict(
            model,
            Xt[va_i],
            A,
            edge_index,
            device
        )

        p_va = apply_platt_scaler(
            p_va_raw,
            cal
        )

        outer_auc = roc_auc_score(
            y[va_i],
            p_va
        )

        oof_pred.append(p_va)
        oof_true.append(y[va_i])
        oof_sample_idx.append(va_i)

        fold_models.append(model)
        fold_calibrators.append(cal)
        histories.append(hist)

        print(
            f"    inner early-stop AUROC "
            f"{inner_auc:.4f} | "
            f"untouched outer AUROC "
            f"{outer_auc:.4f} | "
            f"Platt slope "
            f"{cal['slope']:.3f}"
        )

    # ---------------------------------------------------------
    # Pool untouched outer-fold predictions
    # ---------------------------------------------------------
    oof_pred = np.concatenate(
        oof_pred
    )

    oof_true = np.concatenate(
        oof_true
    )

    threshold = float(
        youden_threshold(
            oof_true,
            oof_pred
        )
    )

    print(
        f"  operating threshold fixed at "
        f"{threshold:.4f} from pooled "
        f"calibrated development OOF "
        f"predictions"
    )

    results = {

        "threshold": threshold,

        "evaluation_protocol":
            "5-fold stratified outer CV; "
            "inner development split for "
            "early stopping and Platt scaling; "
            "pooled calibrated OOF threshold; "
            "frozen calibrated fold ensemble "
            "for external validation",

        "internal": evaluate(
            oof_true,
            oof_pred,
            threshold,
            n_boot=cfg["eval"]["n_bootstrap"],
            label=(
                f"{dev} "
                "(untouched outer-fold OOF CV)"
            )
        ),

        "external": {},

        "predictions": [],

        "fold_calibration":
            fold_calibrators,

        "n_parameters":
            int(
                fold_models[0]
                .n_parameters()
            ),

        "train_seconds":
            float(
                time.time() - t0
            ),
    }

    print(
        f"  internal AUROC "
        f"{results['internal']['auroc']:.4f}"
    )

    # ---------------------------------------------------------
    # Save development OOF predictions
    # ---------------------------------------------------------
    oof_samples = np.concatenate(
        oof_sample_idx
    )

    for s, pi, yi in zip(
        X.index[oof_samples],
        oof_pred,
        oof_true
    ):
        results["predictions"].append(
            {
                "sample": s,
                "cohort": dev,
                "model": "GraphOmics-HCC",
                "p": float(pi),
                "y": int(yi),
                "split":
                    "development_outer_oof",
            }
        )

    # ---------------------------------------------------------
    # Frozen external validation
    # ---------------------------------------------------------
    print(
        "\nexternal validation "
        "(frozen calibrated ensemble, "
        "no refitting)"
    )

    for c in cfg["cohorts"]["external"]:

        idx = np.flatnonzero(
            cohort == c
        )

        if len(idx) == 0:
            print(
                f"  {c}: absent, skipped"
            )
            continue

        if len(np.unique(y[idx])) < 2:
            print(
                f"  {c}: single class, skipped"
            )
            continue

        p = calibrated_ensemble_predict(
            fold_models,
            fold_calibrators,
            Xt[idx],
            A,
            edge_index,
            device
        )

        results["external"][c] = evaluate(
            y[idx],
            p,
            threshold,
            n_boot=(
                cfg["eval"]["n_bootstrap"]
            ),
            label=c
        )

        print(
            f"  {c}: AUROC "
            f"{results['external'][c]['auroc']:.4f} "
            f"("
            f"{results['external'][c]['auroc_ci_low']:.3f}"
            f"–"
            f"{results['external'][c]['auroc_ci_high']:.3f}"
            f"), n={len(idx)}"
        )

        for s, pi, yi in zip(
            X.index[idx],
            p,
            y[idx]
        ):
            results["predictions"].append(
                {
                    "sample": s,
                    "cohort": c,
                    "model":
                        "GraphOmics-HCC",
                    "p": float(pi),
                    "y": int(yi),
                    "split": "external",
                }
            )

    ext = [
        v["auroc"]
        for v in
        results["external"].values()
    ]

    if ext:

        results["external_mean_auroc"] = (
            float(np.mean(ext))
        )

        results["external_auroc_range"] = [
            float(min(ext)),
            float(max(ext))
        ]

        results[
            "internal_external_gap"
        ] = float(
            results["internal"]["auroc"]
            - np.mean(ext)
        )

        print(
            f"\n  mean external AUROC "
            f"{np.mean(ext):.4f} "
            f"(internal-external gap "
            f"{results['internal_external_gap']:+.4f})"
        )

    # ---------------------------------------------------------
    # Save reproducibility artifacts
    # ---------------------------------------------------------
    if save_artifacts:

        torch.save(
            [
                m.state_dict()
                for m in fold_models
            ],
            os.path.join(
                cfg["paths"]["results"],
                "model_weights.pt"
            )
        )

        save_json(
            histories,
            os.path.join(
                cfg["paths"]["results"],
                "training_history.json"
            )
        )

        save_json(
            fold_calibrators,
            os.path.join(
                cfg["paths"]["results"],
                "fold_calibration.json"
            )
        )

    return results, (
        X,
        samples,
        Xt,
        y,
        cohort,
        A,
        edge_index,
        M,
        H,
        fold_models,
        threshold
    )
'''

    text = (
        text[:start]
        + new_run_main
        + text[end:]
    )

    p.write_text(text)

    print("✅ Strict nested/outer-CV evaluation patch applied.")
    print("✅ External-label leakage: none.")
    print("✅ Fold calibration fitted on development data only.")
```

```python
%%writefile src/07_make_publication_figures.py

from pathlib import Path
import json
import textwrap
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch
from matplotlib.ticker import FormatStrFormatter

from sklearn.metrics import (
    roc_curve,
    roc_auc_score,
    precision_recall_curve,
    average_precision_score,
)

warnings.filterwarnings("ignore")

# ============================================================
# PATHS
# ============================================================

ROOT = Path(".")
RESULTS = ROOT / "results"
PROCESSED = ROOT / "data" / "processed"

OUT = RESULTS / "publication_outputs"
FIGDIR = OUT / "figures"
TABLEDIR = OUT / "tables"

FIGDIR.mkdir(parents=True, exist_ok=True)
TABLEDIR.mkdir(parents=True, exist_ok=True)


# ============================================================
# STYLE
# ============================================================

BLUE = "#2457A7"
LIGHT_BLUE = "#5B8FD1"
GREEN = "#2A9D6F"
DARK_GREEN = "#147D64"
ORANGE = "#E79032"
RED = "#C94C4C"
PURPLE = "#775DA6"
GRAY = "#777777"
LIGHT_GRAY = "#E9EDF2"
DARK = "#263238"

COHORT_COLORS = {
    "TCGA-LIHC": "#2457A7",
    "GSE14520": "#2A9D6F",
    "GSE25097": "#E79032",
    "GSE36376": "#775DA6",
    "GSE76427": "#C94C4C",
    "GSE64041": "#5B8FD1",
}

plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "font.size": 10,
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "xtick.labelsize": 9,
    "ytick.labelsize": 9,
    "legend.fontsize": 9,
    "axes.linewidth": 0.8,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "savefig.facecolor": "white",
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})


def save_figure(fig, name):
    png = FIGDIR / f"{name}.png"
    pdf = FIGDIR / f"{name}.pdf"

    fig.savefig(
        png,
        dpi=600,
        bbox_inches="tight",
        facecolor="white"
    )

    fig.savefig(
        pdf,
        bbox_inches="tight",
        facecolor="white"
    )

    plt.close(fig)

    print("✅", png)
    print("✅", pdf)


def clean_axes(ax):
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.grid(
        axis="y",
        alpha=0.15,
        linewidth=0.7
    )


def pretty_pathway(x, width=38):
    x = str(x)

    for prefix in [
        "REACTOME_",
        "KEGG_MEDICUS_REFERENCE_",
        "KEGG_MEDICUS_",
        "KEGG_"
    ]:
        if x.startswith(prefix):
            x = x[len(prefix):]

    x = x.replace("_", " ").title()

    return "\n".join(
        textwrap.wrap(
            x,
            width=width
        )
    )


# ============================================================
# LOAD RESULTS
# ============================================================

main_file = RESULTS / "metrics_main_and_baselines.json"

if not main_file.exists():
    main_file = RESULTS / "metrics_raw.json"

with open(main_file) as f:
    MAIN = json.load(f)

G = MAIN["graphomics"]

baseline_results = MAIN.get(
    "baselines",
    {}
)

ablation_file = (
    RESULTS /
    "metrics_ablations_COMPLETE.json"
)

ABLATIONS = {}

if ablation_file.exists():
    with open(ablation_file) as f:
        ABLATIONS = json.load(f).get(
            "ablations",
            {}
        )

graph_file = RESULTS / "graph_stats.json"

GRAPH = {}

if graph_file.exists():
    with open(graph_file) as f:
        GRAPH = json.load(f)

survival_file = RESULTS / "survival.json"

SURVIVAL = {}

if survival_file.exists():
    with open(survival_file) as f:
        SURVIVAL = json.load(f)

samples = pd.read_csv(
    PROCESSED /
    "sample_table.csv"
)

pred = pd.read_csv(
    RESULTS /
    "predictions_graphomics.csv"
)


# ============================================================
# TABLE 1 — COHORT COMPOSITION
# ============================================================

cohort_table = (
    samples
    .groupby(
        ["cohort", "tumour"]
    )
    .size()
    .unstack(fill_value=0)
)

cohort_table = cohort_table.rename(
    columns={
        0: "Non-tumour",
        1: "Tumour"
    }
)

for col in [
    "Non-tumour",
    "Tumour"
]:
    if col not in cohort_table.columns:
        cohort_table[col] = 0

cohort_table["Total"] = (
    cohort_table["Non-tumour"]
    + cohort_table["Tumour"]
)

cohort_table = (
    cohort_table[
        [
            "Non-tumour",
            "Tumour",
            "Total"
        ]
    ]
    .reset_index()
)

preferred_order = [
    "TCGA-LIHC",
    "GSE14520",
    "GSE25097",
    "GSE36376",
    "GSE76427",
    "GSE64041"
]

cohort_table["order"] = (
    cohort_table["cohort"]
    .map(
        {
            c: i
            for i, c
            in enumerate(preferred_order)
        }
    )
)

cohort_table = (
    cohort_table
    .sort_values("order")
    .drop(columns="order")
)

cohort_table.to_csv(
    TABLEDIR /
    "Table1_Cohort_Composition.csv",
    index=False
)


# ============================================================
# FIGURE 1 — STUDY WORKFLOW
# ============================================================

fig, ax = plt.subplots(
    figsize=(15, 4.8)
)

ax.set_xlim(0, 15)
ax.set_ylim(0, 5)
ax.axis("off")

boxes = [
    (
        0.35, 1.55, 2.0, 1.8,
        "Multi-cohort\ntranscriptomics",
        "TCGA-LIHC +\n5 GEO cohorts",
        BLUE
    ),
    (
        2.75, 1.55, 2.0, 1.8,
        "Preprocessing",
        "QC • gene mapping\nrank normalization",
        LIGHT_BLUE
    ),
    (
        5.15, 1.55, 2.0, 1.8,
        "Biological graph",
        "STRING v12.0\nscore ≥700",
        GREEN
    ),
    (
        7.55, 1.55, 2.0, 1.8,
        "Pathway layer",
        "MSigDB C2:CP\n1,852 pathways",
        DARK_GREEN
    ),
    (
        9.95, 1.55, 2.0, 1.8,
        "GraphOmics-HCC",
        "2-layer GNN\nbiological pooling",
        PURPLE
    ),
    (
        12.35, 1.55, 2.0, 1.8,
        "Validation",
        "5-fold TCGA CV\n5 external cohorts",
        ORANGE
    ),
]

for x, y, w, h, title, subtitle, color in boxes:

    shadow = FancyBboxPatch(
        (x + 0.06, y - 0.06),
        w,
        h,
        boxstyle="round,pad=0.04,rounding_size=0.12",
        facecolor="#D9DEE5",
        edgecolor="none",
        alpha=0.55
    )

    ax.add_patch(shadow)

    box = FancyBboxPatch(
        (x, y),
        w,
        h,
        boxstyle="round,pad=0.04,rounding_size=0.12",
        facecolor="white",
        edgecolor=color,
        linewidth=2.0
    )

    ax.add_patch(box)

    ax.add_patch(
        FancyBboxPatch(
            (x, y + h - 0.43),
            w,
            0.43,
            boxstyle="round,pad=0.04,rounding_size=0.12",
            facecolor=color,
            edgecolor=color
        )
    )

    ax.text(
        x + w / 2,
        y + h - 0.22,
        title,
        ha="center",
        va="center",
        fontsize=10.5,
        fontweight="bold",
        color="white"
    )

    ax.text(
        x + w / 2,
        y + 0.60,
        subtitle,
        ha="center",
        va="center",
        fontsize=9.2,
        color=DARK,
        linespacing=1.35
    )

for i in range(len(boxes) - 1):

    x1 = boxes[i][0] + boxes[i][2]
    x2 = boxes[i + 1][0]

    ax.annotate(
        "",
        xy=(x2 - 0.08, 2.45),
        xytext=(x1 + 0.08, 2.45),
        arrowprops=dict(
            arrowstyle="-|>",
            color="#68737D",
            lw=1.8
        )
    )

ax.text(
    7.5,
    4.42,
    "GraphOmics-HCC: biologically structured multi-cohort HCC classification",
    ha="center",
    fontsize=15,
    fontweight="bold",
    color=DARK
)

ax.text(
    7.5,
    0.58,
    "Explainability • pathway analysis • ablations • exploratory survival analysis",
    ha="center",
    fontsize=10.5,
    color=GRAY
)

save_figure(
    fig,
    "Figure1_GraphOmics_HCC_Workflow"
)


# ============================================================
# FIGURE 2 — COHORT COMPOSITION
# ============================================================

plot_df = cohort_table.copy()

fig, ax = plt.subplots(
    figsize=(9.5, 5.6)
)

x = np.arange(len(plot_df))

non = plot_df["Non-tumour"].values
tum = plot_df["Tumour"].values

ax.bar(
    x,
    non,
    width=0.68,
    label="Non-tumour",
    color=LIGHT_BLUE,
    edgecolor="white",
    linewidth=0.7
)

ax.bar(
    x,
    tum,
    bottom=non,
    width=0.68,
    label="Tumour",
    color=GREEN,
    edgecolor="white",
    linewidth=0.7
)

for i, (a, b) in enumerate(
    zip(non, tum)
):
    ax.text(
        i,
        a / 2,
        str(int(a)),
        ha="center",
        va="center",
        fontsize=8.5,
        color="white",
        fontweight="bold"
    )

    ax.text(
        i,
        a + b / 2,
        str(int(b)),
        ha="center",
        va="center",
        fontsize=8.5,
        color="white",
        fontweight="bold"
    )

    ax.text(
        i,
        a + b + 8,
        f"n={int(a+b)}",
        ha="center",
        fontsize=8.5,
        color=DARK
    )

ax.set_xticks(x)

ax.set_xticklabels(
    plot_df["cohort"],
    rotation=25,
    ha="right"
)

ax.set_ylabel("Number of samples")

ax.set_title(
    "Final analytic cohort composition after quality control",
    fontweight="bold",
    pad=12
)

ax.legend(
    frameon=False,
    ncol=2,
    loc="upper right"
)

clean_axes(ax)

save_figure(
    fig,
    "Figure2_Cohort_Composition"
)


# ============================================================
# FIGURE 3 — ROC CURVES
# ============================================================

fig, ax = plt.subplots(
    figsize=(7.8, 7.0)
)

cohort_order = [
    c for c in preferred_order
    if c in pred["cohort"].unique()
]

for cohort in cohort_order:

    d = pred[
        pred["cohort"] == cohort
    ]

    if d["y"].nunique() < 2:
        continue

    fpr, tpr, _ = roc_curve(
        d["y"],
        d["p"]
    )

    auc_val = roc_auc_score(
        d["y"],
        d["p"]
    )

    label = (
        f"{cohort} "
        f"(AUC={auc_val:.3f})"
    )

    ax.plot(
        fpr,
        tpr,
        lw=2.2,
        label=label,
        color=COHORT_COLORS.get(
            cohort,
            None
        )
    )

ax.plot(
    [0, 1],
    [0, 1],
    "--",
    color="#A7ADB4",
    lw=1.2,
    label="Chance"
)

ax.set_xlim(0, 1)
ax.set_ylim(0, 1.02)

ax.set_xlabel(
    "False-positive rate"
)

ax.set_ylabel(
    "True-positive rate"
)

ax.set_title(
    "GraphOmics-HCC discrimination across development and external cohorts",
    fontweight="bold",
    pad=12
)

ax.legend(
    frameon=False,
    loc="lower right"
)

ax.grid(
    alpha=0.16
)

ax.set_aspect(
    "equal",
    adjustable="box"
)

save_figure(
    fig,
    "Figure3_ROC_Curves"
)


# ============================================================
# SUPPLEMENTARY FIGURE — PR CURVES
# ============================================================

fig, ax = plt.subplots(
    figsize=(7.8, 7.0)
)

for cohort in cohort_order:

    d = pred[
        pred["cohort"] == cohort
    ]

    if d["y"].nunique() < 2:
        continue

    precision, recall, _ = (
        precision_recall_curve(
            d["y"],
            d["p"]
        )
    )

    ap = average_precision_score(
        d["y"],
        d["p"]
    )

    ax.plot(
        recall,
        precision,
        lw=2.2,
        label=(
            f"{cohort} "
            f"(AP={ap:.3f})"
        ),
        color=COHORT_COLORS.get(
            cohort,
            None
        )
    )

ax.set_xlim(0, 1)
ax.set_ylim(0, 1.02)

ax.set_xlabel("Recall")
ax.set_ylabel("Precision")

ax.set_title(
    "Precision–recall performance across cohorts",
    fontweight="bold",
    pad=12
)

ax.legend(
    frameon=False,
    loc="lower left"
)

ax.grid(alpha=0.16)

save_figure(
    fig,
    "FigureS1_Precision_Recall_Curves"
)


# ============================================================
# TABLE 2 — GRAPHOMICS PERFORMANCE
# ============================================================

performance_rows = []

internal = G.get(
    "internal",
    {}
)

tcga_n = int(
    cohort_table.loc[
        cohort_table["cohort"]
        == "TCGA-LIHC",
        "Total"
    ].iloc[0]
)

tcga_pos = int(
    cohort_table.loc[
        cohort_table["cohort"]
        == "TCGA-LIHC",
        "Tumour"
    ].iloc[0]
)

performance_rows.append({
    "Cohort": "TCGA-LIHC",
    "Role": "Development / outer CV",
    "n": tcga_n,
    "Tumour": tcga_pos,
    "AUROC": G.get(
        "internal_cv_mean_auroc",
        internal.get("auroc")
    ),
    "AUROC_CI_low": np.nan,
    "AUROC_CI_high": np.nan,
    "CV_SD": G.get(
        "internal_cv_sd_auroc",
        np.nan
    ),
    "AUPRC": internal.get(
        "auprc",
        np.nan
    ),
    "Sensitivity": internal.get(
        "sensitivity",
        np.nan
    ),
    "Specificity": internal.get(
        "specificity",
        np.nan
    ),
    "Balanced_accuracy": internal.get(
        "balanced_accuracy",
        np.nan
    ),
    "F1": internal.get(
        "f1",
        np.nan
    ),
    "MCC": internal.get(
        "mcc",
        np.nan
    ),
    "Brier": internal.get(
        "brier",
        np.nan
    ),
})

for cohort, m in G.get(
    "external",
    {}
).items():

    performance_rows.append({
        "Cohort": cohort,
        "Role": "External validation",
        "n": m.get("n"),
        "Tumour": m.get("n_pos"),
        "AUROC": m.get("auroc"),
        "AUROC_CI_low": m.get(
            "auroc_ci_low"
        ),
        "AUROC_CI_high": m.get(
            "auroc_ci_high"
        ),
        "CV_SD": np.nan,
        "AUPRC": m.get("auprc"),
        "Sensitivity": m.get(
            "sensitivity"
        ),
        "Specificity": m.get(
            "specificity"
        ),
        "Balanced_accuracy": m.get(
            "balanced_accuracy"
        ),
        "F1": m.get("f1"),
        "MCC": m.get("mcc"),
        "Brier": m.get("brier"),
    })

performance_table = pd.DataFrame(
    performance_rows
)

performance_table.to_csv(
    TABLEDIR /
    "Table2_GraphOmics_Diagnostic_Performance.csv",
    index=False
)


# ============================================================
# FIGURE 4 — EXTERNAL AUROC FOREST
# ============================================================

external = G.get(
    "external",
    {}
)

forest_rows = []

for cohort, m in external.items():

    forest_rows.append({
        "cohort": cohort,
        "auc": m["auroc"],
        "low": m["auroc_ci_low"],
        "high": m["auroc_ci_high"],
        "n": m["n"],
    })

forest = pd.DataFrame(
    forest_rows
).sort_values(
    "auc",
    ascending=True
)

fig, ax = plt.subplots(
    figsize=(8.5, 5.2)
)

y = np.arange(
    len(forest)
)

xerr = np.vstack([
    forest["auc"] - forest["low"],
    forest["high"] - forest["auc"]
])

ax.errorbar(
    forest["auc"],
    y,
    xerr=xerr,
    fmt="o",
    markersize=8,
    color=BLUE,
    ecolor=LIGHT_BLUE,
    elinewidth=2.2,
    capsize=4
)

for i, row in forest.reset_index(
    drop=True
).iterrows():

    ax.text(
        min(
            row["high"] + 0.004,
            1.005
        ),
        i,
        f'{row["auc"]:.3f}',
        va="center",
        fontsize=9
    )

ax.set_yticks(y)

ax.set_yticklabels([
    f'{r.cohort}  (n={r.n})'
    for r in forest.itertuples()
])

ax.set_xlim(
    max(
        0.75,
        forest["low"].min() - 0.03
    ),
    1.01
)

ax.axvline(
    0.5,
    linestyle="--",
    color="#BBBBBB",
    lw=1
)

ax.set_xlabel(
    "AUROC (95% bootstrap CI)"
)

ax.set_title(
    "External validation across five independent HCC cohorts",
    fontweight="bold",
    pad=12
)

ax.grid(
    axis="x",
    alpha=0.18
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_visible(False)

save_figure(
    fig,
    "Figure4_External_AUROC_Forest"
)


# ============================================================
# TABLE 3 + FIGURE 5 — MODEL COMPARISON
# ============================================================

external_cohorts = [
    "GSE14520",
    "GSE25097",
    "GSE36376",
    "GSE76427",
    "GSE64041"
]

comparison_rows = []

row = {
    "Model": "GraphOmics-HCC",
    "Internal_AUROC":
        G.get(
            "internal_cv_mean_auroc",
            internal.get("auroc")
        ),
}

for c in external_cohorts:
    row[c] = (
        G.get("external", {})
        .get(c, {})
        .get("auroc", np.nan)
    )

row["External_mean_AUROC"] = (
    G.get(
        "external_mean_auroc",
        np.nan
    )
)

row["Internal_external_gap"] = (
    G.get(
        "internal_external_gap",
        np.nan
    )
)

comparison_rows.append(row)


for name, b in baseline_results.items():

    r = {
        "Model": name,
        "Internal_AUROC":
            b.get(
                "internal",
                {}
            ).get(
                "auroc",
                np.nan
            ),
    }

    for c in external_cohorts:
        r[c] = (
            b.get(
                "external",
                {}
            ).get(
                c,
                {}
            ).get(
                "auroc",
                np.nan
            )
        )

    r["External_mean_AUROC"] = (
        b.get(
            "external_mean_auroc",
            np.nan
        )
    )

    r["Internal_external_gap"] = (
        b.get(
            "internal_external_gap",
            np.nan
        )
    )

    comparison_rows.append(r)


comparison = pd.DataFrame(
    comparison_rows
)

comparison.to_csv(
    TABLEDIR /
    "Table3_Model_Comparison.csv",
    index=False
)


heat_cols = (
    external_cohorts
    + ["External_mean_AUROC"]
)

heat = comparison.set_index(
    "Model"
)[heat_cols]

fig, ax = plt.subplots(
    figsize=(11.5, 6.0)
)

vals = heat.values.astype(float)

vmin = max(
    0.80,
    np.nanmin(vals) - 0.02
)

im = ax.imshow(
    vals,
    aspect="auto",
    cmap="YlGnBu",
    vmin=vmin,
    vmax=1.0
)

ax.set_xticks(
    np.arange(len(heat_cols))
)

ax.set_xticklabels(
    [
        x.replace(
            "External_mean_AUROC",
            "External mean"
        )
        for x in heat_cols
    ],
    rotation=28,
    ha="right"
)

ax.set_yticks(
    np.arange(len(heat.index))
)

ax.set_yticklabels(
    heat.index
)

for i in range(vals.shape[0]):
    for j in range(vals.shape[1]):

        if np.isnan(vals[i, j]):
            txt = "NA"
        else:
            txt = f"{vals[i,j]:.3f}"

        ax.text(
            j,
            i,
            txt,
            ha="center",
            va="center",
            fontsize=8.5,
            color=(
                "white"
                if (
                    not np.isnan(vals[i,j])
                    and vals[i,j] >
                    (vmin + 1) / 2
                )
                else DARK
            ),
            fontweight=(
                "bold"
                if i == 0
                else "normal"
            )
        )

cbar = fig.colorbar(
    im,
    ax=ax,
    fraction=0.032,
    pad=0.03
)

cbar.set_label(
    "AUROC"
)

ax.set_title(
    "External validation performance of GraphOmics-HCC and conventional baselines",
    fontweight="bold",
    pad=14
)

for spine in ax.spines.values():
    spine.set_visible(False)

save_figure(
    fig,
    "Figure5_Model_Comparison_Heatmap"
)


# ============================================================
# TABLE 4 + FIGURE 6 — ABLATIONS
# ============================================================

ablation_rows = []

full_external = G.get(
    "external_mean_auroc",
    np.nan
)

full_internal = G.get(
    "internal_cv_mean_auroc",
    internal.get(
        "auroc",
        np.nan
    )
)

for name, m in ABLATIONS.items():

    if "error" in m:
        continue

    ext_auc = m.get(
        "external_mean_auroc",
        np.nan
    )

    int_auc = m.get(
        "internal_auroc",
        np.nan
    )

    ablation_rows.append({
        "Variant": name,
        "Internal_AUROC":
            int_auc,
        "External_mean_AUROC":
            ext_auc,
        "Delta_external_vs_full":
            ext_auc - full_external,
        "Internal_external_gap":
            m.get(
                "internal_external_gap",
                np.nan
            ),
    })


ablation_table = pd.DataFrame(
    ablation_rows
)

ablation_table.to_csv(
    TABLEDIR /
    "Table4_Ablation_Analysis.csv",
    index=False
)


if len(ablation_table):

    plot = (
        ablation_table
        .sort_values(
            "Delta_external_vs_full"
        )
        .copy()
    )

    fig, ax = plt.subplots(
        figsize=(9.5, 6.0)
    )

    colors = [
        GREEN if x >= 0
        else BLUE
        for x in
        plot["Delta_external_vs_full"]
    ]

    bars = ax.barh(
        plot["Variant"],
        plot[
            "Delta_external_vs_full"
        ],
        color=colors,
        alpha=0.9
    )

    ax.axvline(
        0,
        color=DARK,
        lw=1.2
    )

    for bar, val in zip(
        bars,
        plot[
            "Delta_external_vs_full"
        ]
    ):

        ax.text(
            val + (
                0.00015
                if val >= 0
                else -0.00015
            ),
            bar.get_y()
            + bar.get_height()/2,
            f"{val:+.4f}",
            ha=(
                "left"
                if val >= 0
                else "right"
            ),
            va="center",
            fontsize=8.5
        )

    ax.set_xlabel(
        "Change in mean external AUROC vs full GraphOmics-HCC"
    )

    ax.set_title(
        "Ablation analysis: effect of individual architectural components",
        fontweight="bold",
        pad=12
    )

    ax.grid(
        axis="x",
        alpha=0.15
    )

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_visible(False)

    save_figure(
        fig,
        "Figure6_Ablation_Delta_AUROC"
    )


# ============================================================
# TABLE 5 — GRAPH CHARACTERISTICS
# ============================================================

graph_labels = {
    "n_genes":
        "Gene nodes",
    "n_edges":
        "STRING edges",
    "mean_degree":
        "Mean degree",
    "median_degree":
        "Median degree",
    "max_degree":
        "Maximum degree",
    "n_isolated":
        "Isolated genes",
    "largest_component_fraction":
        "Largest component fraction",
    "n_components":
        "Connected components",
    "n_pathways":
        "MSigDB pathways",
    "pathway_density":
        "Gene-pathway mask density",
    "genes_per_pathway_mean":
        "Mean genes per pathway",
    "pathways_per_gene_mean":
        "Mean pathways per gene",
    "gene_pathway_coverage":
        "Gene pathway coverage",
    "string_cutoff":
        "STRING score threshold",
    "string_version":
        "STRING version",
    "msigdb_collection":
        "MSigDB collection",
}

graph_rows = []

for key, label in graph_labels.items():

    if key in GRAPH:
        graph_rows.append({
            "Characteristic": label,
            "Value": GRAPH[key]
        })

graph_table = pd.DataFrame(
    graph_rows
)

graph_table.to_csv(
    TABLEDIR /
    "Table5_Graph_Characteristics.csv",
    index=False
)


# ============================================================
# FIGURE 7 — EXPLAINABILITY
# ============================================================

gene_file = (
    RESULTS /
    "gene_attributions_full.csv"
)

shap_file = (
    RESULTS /
    "pathway_shap.csv"
)

if (
    gene_file.exists()
    and shap_file.exists()
):

    genes = pd.read_csv(
        gene_file
    )

    pathways = pd.read_csv(
        shap_file
    )

    top_genes = (
        genes
        .sort_values(
            "abs_attribution",
            ascending=False
        )
        .head(15)
        .sort_values(
            "abs_attribution",
            ascending=True
        )
    )

    top_pathways = (
        pathways
        .sort_values(
            "mean_abs_shap",
            ascending=False
        )
        .head(15)
        .sort_values(
            "mean_abs_shap",
            ascending=True
        )
    )

    fig, axes = plt.subplots(
        1,
        2,
        figsize=(15, 8)
    )

    # --------------------------------------------------------
    # Gene attribution
    # --------------------------------------------------------

    gene_colors = [
        GREEN
        if x >= 0
        else BLUE
        for x in
        top_genes[
            "mean_attribution"
        ]
    ]

    axes[0].barh(
        top_genes["gene"],
        top_genes[
            "mean_attribution"
        ],
        color=gene_colors,
        alpha=0.92
    )

    axes[0].axvline(
        0,
        color=DARK,
        lw=1
    )

    axes[0].set_xlabel(
        "Mean integrated-gradient attribution"
    )

    axes[0].set_title(
        "A  Top gene-level attributions",
        loc="left",
        fontweight="bold"
    )

    axes[0].grid(
        axis="x",
        alpha=0.15
    )

    axes[0].spines[
        "top"
    ].set_visible(False)

    axes[0].spines[
        "right"
    ].set_visible(False)


    # --------------------------------------------------------
    # Pathway SHAP
    # --------------------------------------------------------

    names = [
        pretty_pathway(
            x,
            34
        )
        for x in
        top_pathways["pathway"]
    ]

    axes[1].barh(
        names,
        top_pathways[
            "mean_abs_shap"
        ],
        color=PURPLE,
        alpha=0.9
    )

    axes[1].set_xlabel(
        "Mean absolute SHAP value"
    )

    axes[1].set_title(
        "B  Pathway-level importance",
        loc="left",
        fontweight="bold"
    )

    axes[1].grid(
        axis="x",
        alpha=0.15
    )

    axes[1].spines[
        "top"
    ].set_visible(False)

    axes[1].spines[
        "right"
    ].set_visible(False)

    fig.suptitle(
        "GraphOmics-HCC biological interpretation",
        fontsize=15,
        fontweight="bold",
        y=1.01
    )

    fig.tight_layout()

    save_figure(
        fig,
        "Figure7_Explainability_Genes_Pathways"
    )

    genes.head(50).to_csv(
        TABLEDIR /
        "TableS1_Top50_Gene_Attributions.csv",
        index=False
    )

    pathways.head(50).to_csv(
        TABLEDIR /
        "TableS2_Top50_Pathway_SHAP.csv",
        index=False
    )


# ============================================================
# TABLE S3 — EXPLAINABILITY STABILITY
# ============================================================

stab_file = (
    RESULTS /
    "explanation_stability.json"
)

if stab_file.exists():

    with open(stab_file) as f:
        explain = json.load(f)

    stability_rows = []

    stability = explain.get(
        "stability",
        {}
    )

    for key, value in stability.items():

        if isinstance(
            value,
            dict
        ):
            row = {
                "Metric": key
            }

            row.update(value)

            stability_rows.append(
                row
            )

        else:
            stability_rows.append({
                "Metric": key,
                "Value": value
            })

    pd.DataFrame(
        stability_rows
    ).to_csv(
        TABLEDIR /
        "TableS3_Explanation_Stability.csv",
        index=False
    )

    sanity = explain.get(
        "sanity_checks",
        {}
    )

    pd.DataFrame(
        [
            {
                "Check": k,
                "Value": v
            }
            for k, v
            in sanity.items()
            if isinstance(
                v,
                (int, float)
            )
        ]
    ).to_csv(
        TABLEDIR /
        "TableS4_Explanation_Sanity_Checks.csv",
        index=False
    )


# ============================================================
# FIGURE 8 — KAPLAN–MEIER
# ============================================================

curve_file = (
    RESULTS /
    "survival_curves.csv"
)

if (
    curve_file.exists()
    and SURVIVAL
):

    curves = pd.read_csv(
        curve_file
    )

    tcga_surv = (
        SURVIVAL
        .get(
            "cohorts",
            {}
        )
        .get(
            "TCGA-LIHC",
            {}
        )
    )

    fig, ax = plt.subplots(
        figsize=(8.0, 6.3)
    )

    group_settings = {
        "low": (
            BLUE,
            "Low diagnostic score"
        ),
        "high": (
            RED,
            "High diagnostic score"
        )
    }

    for group in [
        "low",
        "high"
    ]:

        d = curves[
            curves["group"]
            == group
        ].sort_values(
            "time"
        )

        if len(d) == 0:
            continue

        color, label = (
            group_settings[group]
        )

        ax.step(
            d["time"],
            d["survival"],
            where="post",
            lw=2.5,
            color=color,
            label=label
        )

    ax.set_xlabel(
        "Overall survival (months)"
    )

    ax.set_ylabel(
        "Survival probability"
    )

    ax.set_ylim(
        0,
        1.02
    )

    ax.set_title(
        "Exploratory survival stratification in TCGA-LIHC",
        fontweight="bold",
        pad=12
    )

    txt = (
        f'n={tcga_surv.get("n", "NA")}, '
        f'events={tcga_surv.get("n_events", "NA")}\n'
        f'Log-rank p={tcga_surv.get("logrank_p", np.nan):.3g}\n'
        f'Median OS: '
        f'{tcga_surv.get("median_os_low", np.nan):.1f} vs '
        f'{tcga_surv.get("median_os_high", np.nan):.1f} months'
    )

    ax.text(
        0.97,
        0.96,
        txt,
        transform=ax.transAxes,
        ha="right",
        va="top",
        fontsize=9.5,
        bbox=dict(
            boxstyle="round,pad=0.5",
            facecolor="white",
            edgecolor="#CDD3D8",
            alpha=0.95
        )
    )

    ax.legend(
        frameon=False,
        loc="lower left"
    )

    ax.grid(
        alpha=0.15
    )

    clean_axes(ax)

    save_figure(
        fig,
        "Figure8_Kaplan_Meier_Survival"
    )


# ============================================================
# TABLE 6 — SURVIVAL
# ============================================================

survival_rows = []

for cohort, r in (
    SURVIVAL
    .get(
        "cohorts",
        {}
    )
    .items()
):

    if "skipped" in r:
        continue

    survival_rows.append({
        "Cohort": cohort,
        "n": r.get("n"),
        "Events": r.get(
            "n_events"
        ),
        "High_group_n": r.get(
            "n_high"
        ),
        "Low_group_n": r.get(
            "n_low"
        ),
        "Cutpoint": r.get(
            "cutpoint"
        ),
        "Logrank_p": r.get(
            "logrank_p"
        ),
        "Median_OS_high_months":
            r.get(
                "median_os_high"
            ),
        "Median_OS_low_months":
            r.get(
                "median_os_low"
            ),
        "HR_univariable":
            r.get(
                "hr_univariable"
            ),
        "HR_univariable_CI_low":
            (
                r.get(
                    "hr_univariable_ci",
                    [np.nan, np.nan]
                )[0]
            ),
        "HR_univariable_CI_high":
            (
                r.get(
                    "hr_univariable_ci",
                    [np.nan, np.nan]
                )[1]
            ),
        "HR_adjusted":
            r.get(
                "hr_adjusted"
            ),
        "HR_adjusted_CI_low":
            (
                r.get(
                    "hr_adjusted_ci",
                    [np.nan, np.nan]
                )[0]
            ),
        "HR_adjusted_CI_high":
            (
                r.get(
                    "hr_adjusted_ci",
                    [np.nan, np.nan]
                )[1]
            ),
        "Adjusted_p":
            r.get(
                "hr_adjusted_p"
            ),
        "C_index_score_only":
            r.get(
                "c_index"
            ),
        "C_index_clinical_only":
            r.get(
                "c_index_clinical_only"
            ),
        "C_index_adjusted":
            r.get(
                "c_index_adjusted"
            ),
        "Delta_C_index":
            r.get(
                "delta_c_index"
            ),
    })


survival_table = pd.DataFrame(
    survival_rows
)

survival_table.to_csv(
    TABLEDIR /
    "Table6_Exploratory_Survival.csv",
    index=False
)


# ============================================================
# EXCEL WORKBOOK
# ============================================================

excel_path = (
    OUT /
    "GraphOmics_HCC_Publication_Tables.xlsx"
)

tables = {
    "Table1_Cohorts":
        cohort_table,
    "Table2_Performance":
        performance_table,
    "Table3_Baselines":
        comparison,
    "Table4_Ablations":
        ablation_table,
    "Table5_Graph":
        graph_table,
    "Table6_Survival":
        survival_table,
}

if gene_file.exists():
    tables[
        "TableS1_Genes"
    ] = pd.read_csv(
        gene_file
    ).head(50)

if shap_file.exists():
    tables[
        "TableS2_Pathways"
    ] = pd.read_csv(
        shap_file
    ).head(50)


with pd.ExcelWriter(
    excel_path,
    engine="openpyxl"
) as writer:

    for sheet, df in tables.items():

        df.to_excel(
            writer,
            sheet_name=sheet[:31],
            index=False
        )


# ============================================================
# FORMAT EXCEL
# ============================================================

try:

    from openpyxl import load_workbook
    from openpyxl.styles import (
        PatternFill,
        Font,
        Alignment,
        Border,
        Side
    )

    wb = load_workbook(
        excel_path
    )

    header_fill = PatternFill(
        "solid",
        fgColor="2457A7"
    )

    header_font = Font(
        color="FFFFFF",
        bold=True
    )

    thin = Side(
        style="thin",
        color="D9DEE5"
    )

    border = Border(
        bottom=thin
    )

    for ws in wb.worksheets:

        ws.freeze_panes = "A2"
        ws.auto_filter.ref = (
            ws.dimensions
        )

        for cell in ws[1]:

            cell.fill = header_fill
            cell.font = header_font

            cell.alignment = Alignment(
                horizontal="center",
                vertical="center",
                wrap_text=True
            )

        for row in ws.iter_rows(
            min_row=2
        ):

            for cell in row:

                cell.border = border

                if isinstance(
                    cell.value,
                    float
                ):

                    cell.number_format = (
                        "0.0000"
                    )

        for column_cells in ws.columns:

            letter = (
                column_cells[0]
                .column_letter
            )

            width = 0

            for cell in column_cells:

                if cell.value is not None:

                    width = max(
                        width,
                        len(
                            str(
                                cell.value
                            )
                        )
                    )

            ws.column_dimensions[
                letter
            ].width = min(
                width + 2,
                38
            )

    wb.save(
        excel_path
    )

except Exception as exc:

    print(
        "Excel formatting warning:",
        exc
    )


# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("PUBLICATION OUTPUTS COMPLETE")
print("=" * 80)

print("\nFigures:")
for f in sorted(
    FIGDIR.glob("*")
):
    print(" ", f)

print("\nTables:")
for f in sorted(
    TABLEDIR.glob("*")
):
    print(" ", f)

print("\nExcel workbook:")
print(" ", excel_path)
```
